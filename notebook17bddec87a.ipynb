```python
import pandas as pd
import numpy as np

# Load country metadata
metadata_path = '/kaggle/input/datasets/amimamoraa/malaria/Metadata_Country_API_SH.MLR.INCD.P3_DS2_en_csv_v2_572.csv'

metadata = pd.read_csv(metadata_path)

# Select the study period
years = [str(year) for year in range(2000, 2025)]

# Convert annual observations to numeric
malaria[years] = malaria[years].apply(
    pd.to_numeric, errors='coerce'
)

# Attach country regions and income groups
df = malaria.merge(
    metadata[['Country Code', 'Region', 'IncomeGroup']],
    on='Country Code',
    how='left'
)

# Exclude regional aggregates and retain actual countries
country_data = df[df['Region'].notna()].copy()

# Retain countries with complete observations for all 25 years
complete_countries = country_data.dropna(
    subset=years
).copy()

print("Countries with complete data:", len(complete_countries))
print("Study period: 2000–2024")
print("Total country-year observations:", len(complete_countries) * 25)

# Calculate annual averages
annual_means = complete_countries[years].mean()

print("\nAnnual averages:")
display(annual_means.to_frame(name='Average_Malaria_Incidence'))
```

```python
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Calculate annual averages
x = np.array([int(year) for year in years])
y = annual_means.values

# Calculate trend
trend = linregress(x, y)
trend_line = trend.intercept + trend.slope * x

# Calculate percentage decline
decline = ((y[0] - y[-1]) / y[0]) * 100

plt.figure(figsize=(12, 6))

plt.plot(x, y, marker='o', linewidth=2.5,
         label='Average malaria incidence')

plt.plot(x, trend_line, linestyle='--',
         linewidth=1.8, label='Linear trend')

plt.annotate(f'{y[0]:.2f}', (x[0], y[0]),
             xytext=(8, 12), textcoords='offset points')

plt.annotate(f'{y[-1]:.2f}', (x[-1], y[-1]),
             xytext=(-45, 12), textcoords='offset points')

plt.title(
    'Malaria Incidence Trends Across 97 Countries (2000–2024)',
    fontsize=15, fontweight='bold'
)
plt.xlabel('Year')
plt.ylabel('Average incidence per 1,000 population at risk')
plt.xticks(range(2000, 2025, 2), rotation=45)
plt.grid(alpha=0.25)
plt.legend()

plt.figtext(
    0.5, 0.01,
    f'Average incidence declined by {decline:.2f}% | '
    f'97 countries | 2,425 observations',
    ha='center', fontsize=10
)

plt.tight_layout(rect=[0, 0.05, 1, 1])

plt.savefig(
    '/kaggle/working/malaria_trend_linkedin.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()

print("Percentage decline:", round(decline, 2), "%")
print("Trend slope:", round(trend.slope, 4))
print("R-squared:", round(trend.rvalue**2, 4))
```

```python
from scipy.stats import kruskal

# Calculate each country's average malaria incidence
country_income_averages = complete_countries.copy()

country_income_averages['Average_Incidence'] = (
    country_income_averages[years].mean(axis=1)
)

# Separate countries by income group
income_groups = [
    group['Average_Incidence'].values
    for name, group in country_income_averages.groupby('IncomeGroup')
]

# Conduct Kruskal-Wallis test
h_statistic, p_value = kruskal(*income_groups)

print("KRUSKAL-WALLIS TEST: INCOME GROUPS")
print("-----------------------------------")

print("H-statistic:", round(h_statistic, 4))
print("P-value:", p_value)

if p_value < 0.05:
    print("Conclusion: Significant differences exist between income groups.")
else:
    print("Conclusion: No statistically significant differences detected.")
```

```python
from scipy.stats import mannwhitneyu
from itertools import combinations
from statsmodels.stats.multitest import multipletests
import pandas as pd

# Prepare country-level average incidence
country_income_averages = complete_countries.copy()

country_income_averages['Average_Incidence'] = (
    country_income_averages[years].mean(axis=1)
)

# Get income group names
groups = country_income_averages['IncomeGroup'].dropna().unique()

results = []

# Compare every pair of income groups
for group1, group2 in combinations(groups, 2):

    values1 = country_income_averages.loc[
        country_income_averages['IncomeGroup'] == group1,
        'Average_Incidence'
    ]

    values2 = country_income_averages.loc[
        country_income_averages['IncomeGroup'] == group2,
        'Average_Incidence'
    ]

    statistic, p_value = mannwhitneyu(
        values1,
        values2,
        alternative='two-sided'
    )

    results.append({
        'Income_Group_1': group1,
        'Income_Group_2': group2,
        'U_statistic': statistic,
        'Raw_P_value': p_value
    })

# Create results table
pairwise_results = pd.DataFrame(results)

# Holm correction for multiple comparisons
pairwise_results['Holm_Adjusted_P'] = multipletests(
    pairwise_results['Raw_P_value'],
    method='holm'
)[1]

pairwise_results['Significant'] = (
    pairwise_results['Holm_Adjusted_P'] < 0.05
)

print("PAIRWISE INCOME-GROUP COMPARISONS")
display(pairwise_results.round(4))
```

```python
# UPDATED REGIONAL MALARIA TRENDS (2000–2024)

regional_trends = (
    complete_countries
    .groupby('Region')[years]
    .mean()
)

regional_changes = pd.DataFrame({
    'Incidence_2000': regional_trends['2000'],
    'Incidence_2024': regional_trends['2024']
})

regional_changes['Absolute_Change'] = (
    regional_changes['Incidence_2024']
    - regional_changes['Incidence_2000']
)

regional_changes['Percentage_Change'] = (
    regional_changes['Absolute_Change']
    / regional_changes['Incidence_2000']
) * 100

print("UPDATED REGIONAL MALARIA CHANGES (2000–2024)")
display(regional_changes.sort_values(
    by='Incidence_2000',
    ascending=False
))
```

```python
# UPDATED REGIONAL LINEAR REGRESSION (2000–2024)

from scipy.stats import linregress

regional_regression = []

for region, row in regional_trends.iterrows():

    result = linregress(
        range(2000, 2025),
        row.values
    )

    regional_regression.append({
        'Region': region,
        'Slope': result.slope,
        'R_squared': result.rvalue ** 2,
        'P_value': result.pvalue
    })

regional_regression = pd.DataFrame(
    regional_regression
).sort_values(
    by='Slope'
)

print("UPDATED REGIONAL REGRESSION RESULTS")
display(regional_regression)
```

```python
# FINAL UPDATED REGIONAL MALARIA TREND CHART

import matplotlib.pyplot as plt

plt.figure(figsize=(14, 7))

for region, row in regional_trends.iterrows():
    plt.plot(
        years,
        row.values,
        marker='o',
        markersize=3,
        label=region
    )

plt.title('Regional Malaria Incidence Trends (2000–2024)')
plt.xlabel('Year')
plt.ylabel('Malaria Incidence per 1,000 Population at Risk')

plt.xticks(range(0, 25, 2), rotation=45)

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# VERIFY INCOME-GROUP PAIRWISE COMPARISONS

from itertools import combinations
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests
import pandas as pd

income_groups = [
    'Low income',
    'Lower middle income',
    'Upper middle income',
    'High income'
]

pairwise_results = []

for group1, group2 in combinations(income_groups, 2):

    data1 = country_changes.loc[
        country_changes['IncomeGroup'] == group1,
        'Average_Incidence'
    ].dropna()

    data2 = country_changes.loc[
        country_changes['IncomeGroup'] == group2,
        'Average_Incidence'
    ].dropna()

    statistic, p_value = mannwhitneyu(
        data1,
        data2,
        alternative='two-sided'
    )

    pairwise_results.append({
        'Group_1': group1,
        'Group_2': group2,
        'U_statistic': statistic,
        'Raw_p_value': p_value
    })

pairwise_results = pd.DataFrame(pairwise_results)

pairwise_results['Holm_adjusted_p'] = multipletests(
    pairwise_results['Raw_p_value'],
    method='holm'
)[1]

pairwise_results['Significant'] = (
    pairwise_results['Holm_adjusted_p'] < 0.05
)

display(pairwise_results)
```
